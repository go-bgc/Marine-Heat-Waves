```python
boto_session = boto3.Session()
s3_creds = boto_session.get_credentials()
s3_creds2 = s3_creds.get_frozen_credentials()

print(s3_creds2)
```

```python
fs = s3fs.S3FileSystem(anon=False, client_kwargs={"region_name":"us-west-2"})
```

```python
local_path = "/Users/steviewalker/Documents/GO-BGC_MHW/GOBAI-O2-v2.3_oxysat.nc"
s3_path = "s3://uw-escience-scratch-prod/team_mhw/GOBAI-O2-v2.3_oxysat.nc"

fs.put(local_path, s3_path, recursive=True)
print(f"Done: {s3_path}")
```
